```python
print("Распределение Pclass:")
print(df['Pclass'].value_counts().sort_index())

print("\nРаспределение Pclass по полу:")
print(pd.crosstab(df['Pclass'], df['Sex']))

print("\nМужчин 2-го класса:", df[(df['Pclass'] == 2) & (df['Sex'] == 'male')].shape[0])
```

```python
print("Медиана Fare:", round(df['Fare'].median(), 2))
print("СКО Fare:", round(df['Fare'].std(), 2))
```

```python
young = df[df['Age'] < 30]
old = df[df['Age'] > 60]
print("Доля выживших моложе 30:", round(young['Survived'].mean(), 4))
print("Доля выживших старше 60:", round(old['Survived'].mean(), 4))
print("Моложе 30 выживали чаще:", young['Survived'].mean() > old['Survived'].mean())
```

```python
print("Доля выживших женщин:", round(df[df['Sex'] == 'female']['Survived'].mean(), 4))
print("Доля выживших мужчин:", round(df[df['Sex'] == 'male']['Survived'].mean(), 4))
print("Женщины выживали чаще:", df[df['Sex'] == 'female']['Survived'].mean() > df[df['Sex'] == 'male']['Survived'].mean())
```

```python
male_names = df[df['Sex'] == 'male']['Name'].str.extract(r'^(?:[^,]+, )?(?:Mr\. |Master\. |Dr\. |Rev\. |Major\. |Col\. |Capt\. |Don\. )?([A-Z][a-z]+)')[0]
print("Самое популярное мужское имя:")
print(male_names.value_counts().head(1))
```

```python
pivot = df.pivot_table(values='Age', index='Pclass', columns='Sex', aggfunc='mean')
print(pivot.round(2))

m1 = pivot.loc[1, 'male']
f1 = pivot.loc[1, 'female']
print("\nМужчины 1 класса старше 40:", m1 > 40)
print("Женщины 1 класса старше 40:", f1 > 40)
print("Мужчины всех классов старше женщин того же класса:",
      (pivot['male'] > pivot['female']).all())
print("Средний возраст по классам убывает (1 > 2 > 3):",
      (pivot.mean(axis=1).diff().dropna() < 0).all())
```

```python
print("Средний возраст погибших:", round(df[df['Survived'] == 0]['Age'].mean(), 2))
print("Средний возраст спасенных:", round(df[df['Survived'] == 1]['Age'].mean(), 2))
print("Средний возраст погибших выше:",
      df[df['Survived'] == 0]['Age'].mean() > df[df['Survived'] == 1]['Age'].mean())
```
