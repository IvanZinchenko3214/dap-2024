```python
import random

array = [[random.randint(1, 100) for _ in range(10)] for _ in range(10)]

for row in array:
    print(' '.join(f'{x:3}' for x in row))

minimum = min(min(row) for row in array)
maximum = max(max(row) for row in array)

print(f'\nМинимум: {minimum}')
print(f'Максимум: {maximum}')
```

```python

matrix = [[(i + j) % 2 for j in range(8)] for i in range(8)]

for row in matrix:
    print(' '.join(map(str, row)))
```

```python
import numpy as np

z = np.arange(11)

z[(z > 3) & (z < 8)] *= -1

print(z)
```

```python
import numpy as np

z = np.random.randint(-50, 51, size=(5, 10))

print("Исходная матрица:")
print(z)

result = z - z.mean(axis=1, keepdims=True)

print("\nРезультат (среднее каждой строки вычтено):")
print(result)

print("\nСредние значения строк результата:")
print(result.mean(axis=1))
```

```python
import numpy as np

z = np.arange(25).reshape(5, 5)

print("Исходная матрица:")
print(z)

z[[0, 1]] = z[[1, 0]]

print("\nМатрица после обмена 1-й и 2-й строк:")
print(z)
```

```python
import numpy as np

z = np.random.random((10, 2))

x = z[:, 0]
y = z[:, 1]

r = np.sqrt(x**2 + y**2)
phi = np.arctan2(y, x)

polar = np.column_stack((r, phi))

print("Декартовы координаты (x, y):")
print(z)

print("\nПолярные координаты (r, phi):")
print(polar)
```

```python
import numpy as np

z = np.random.random(10)

print("Исходный массив:")
print(z)

z[z.argmax()] = 0

print("\nМассив после замены максимума на 0:")
print(z)
```

```python
import numpy as np

z = np.ones(10)
i = np.random.randint(0, len(z), 20)

print("Индексы (могут повторяться):")
print(i)
print("Исходный ммассив")
print(z)
np.add.at(z, i, 1)

print("\nРезультат:")
print(z)
```

```python
import numpy as np

z = np.array([1, 2, 3, 4, 5])

result = np.zeros((len(z) - 1) * 4 + 1, dtype=z.dtype)

result[::4] = z

print(result)
```

```python
import numpy as np

z = np.arange(100)
v = np.random.uniform(0, 100)

print(f"Искомое значение: {v}")
print(f"Массив:\n {z}")

index = np.abs(z - v).argmin()

print(f"Ближайшее число: {z[index]}")
print(f"Индекс: {index}")
```
