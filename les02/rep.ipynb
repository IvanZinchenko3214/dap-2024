```python
def twoSumBrute(nums, target):
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return [i, j]
    return []

nums = [2, 7, 11, 15, 1, -5, 30, -2]
target = -1

result = twoSumBrute(nums, target)
print(result)
```

```python
def isPalindrome(x: int) -> bool:

    if x < 0:
        return False

    str_x = str(x)

    return str_x == str_x[::-1]

exmpl = 2112
result = isPalindrome(exmpl)
print(result)
```

```python
def romanToInt(s: str) -> int:

    values = {
        'I': 1,
        'V': 5,
        'X': 10,
        'L': 50,
        'C': 100,
        'D': 500,
        'M': 1000
    }

    total = 0
    n = len(s)

    for i in range(n):
        current = values[s[i]]

        if i + 1 < n and current < values[s[i + 1]]:
            total -= current
        else:
            total += current

    return total


result = romanToInt("XL")
print(result)
```

```python

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def mergeTwoLists(list1, list2):
    if not list1:
        return list2
    if not list2:
        return list1

    if list1.val <= list2.val:
        list1.next = mergeTwoLists(list1.next, list2)
        return list1
    else:
        list2.next = mergeTwoLists(list1, list2.next)
        return list2

def create_list(arr):
    dummy = ListNode(0)
    current = dummy
    for val in arr:
        current.next = ListNode(val)
        current = current.next
    return dummy.next

def list_to_array(head):
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result

arr1 = [1, 2, 4]
arr2 = [1, 3, 4]

list1 = create_list(arr1)
list2 = create_list(arr2)

result = mergeTwoLists(list1, list2)

print(list_to_array(result))
```

```python
def longestCommonPrefix(strs):
    if not strs:
        return ""

    for i in range(len(strs[0])):
        char = strs[0][i]
        for j in range(1, len(strs)):
            if i >= len(strs[j]) or strs[j][i] != char:
                return strs[0][:i]

    return strs[0]


test_cases = ["abc", "abcd", "ab"]

result = longestCommonPrefix(test_cases)

print(f"longestCommonPrefix({test_cases}) = '{result}'")
```

```python
def lengthOfLongestSubstring(s: str) -> int:

    def all_unique(substring):
        return len(set(substring)) == len(substring)

    n = len(s)
    max_length = 0

    for i in range(n):
        for j in range(i, n):
            if all_unique(s[i:j+1]):
                max_length = max(max_length, j - i + 1)

    return max_length

result = lengthOfLongestSubstring("pwwkew")
print(result)
```
